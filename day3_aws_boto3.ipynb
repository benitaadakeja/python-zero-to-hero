```python
import boto3
from botocore.exceptions import NoCredentialsError
s3_client = boto3.client('s3')
response = s3_client.list_buckets()
for bucket in response['Buckets']:
    print(f"Found bucket: {bucket['Name']}")
#try:
 #   response = s3_client.list_buckets()
 #   print("Success: Connected to AWS!")
#except NoCredentialsError:
#    print(f"SECURITY ALERT: AWS credentials missing. Access Denied to {s3_client.meta.service_model.service_name}.")
```

```python
import boto3
s3_client = boto3.client('s3')
new_bucket_name = "python-zero-to-hero-boto3-457909"
response = s3_client.create_bucket(Bucket=new_bucket_name)
print(f"Success! Provisioned new bucket: {new_bucket_name}")
```

```python
s3_client.upload_file(
    Filename = "login_audit.txt",
    Bucket = "python-zero-to-hero-boto3-457909",
    Key = "login_audit.txt"
)
print("Upload complete! login_audit.txt is now in s3")
```

```python
response = s3_client.list_objects_v2(Bucket="python-zero-to-hero-boto3-457909")
print("-------------Files in my Bucket----------------")
if 'Contents'in response:
    for obj in response['Contents']:
        print(f"Found file: {obj['Key']}")
else:
    print("The bucket is currently empty.")
```

```python
# Step 1: Delete the file (object) inside the bucket
s3_client.delete_object(
    Bucket="python-zero-to-hero-boto3-457909",
    Key="login_audit.txt"
)
print("File deleted.")

# Step 2: Delete the now-empty bucket
s3_client.delete_bucket(
    Bucket="python-zero-to-hero-boto3-457909"
)
print("Bucket deleted. Infrastructure successfully torn down!")
```
